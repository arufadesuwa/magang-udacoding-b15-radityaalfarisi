## 1. Variabel, Tipe Data, Casting, dan f-string

- **Variabel** digunakan untuk menyimpan data di memori.
- **Tipe data primitif** di Python meliputi `str` (teks), `int` (bilangan bulat), `float` (bilangan desimal), dan `bool` (boolean `True`/`False`).
- **Casting** adalah proses mengubah tipe data dari satu tipe ke tipe lainnya menggunakan fungsi seperti `int()`, `float()`, `str()`, dan `bool()`.
- **f-string** (`f"...{variabel}..."`) memudahkan penyisipan nilai variabel langsung ke dalam string.

```python
# Deklarasi variabel dengan type hint
name: str = "Raditya Alfarisi"
age: int = 17
tall: float = 167.1
is_graduate: bool = False

# Menampilkan dengan f-string
print(f"Nama  : {name} (tipe: {type(name).__name__})")
print(f"Umur  : {age} tahun (tipe: {type(age).__name__})")
print(f"Tinggi: {tall} cm (tipe: {type(tall).__name__})")
print(f"Lulus?: {is_graduate} (tipe: {type(is_graduate).__name__})")

# Contoh Casting (konversi tipe data)
angka_float = float("85.5") # str jadi float
age_str = str(age)          # int jadi str
tall_int = int(tall)        # float jadi int (angka belakang koma diilangin)

print(f"\nHasil casting")
print(f"angka_float: {angka_float}")
print(f"age_str    : '{age_str}'")
print(f"tall_int   : {tall_int}")
```

## 2. Operator & Input/Output

- **Operator aritmatika**: `+`, `-`, `*`, `/` (pembagian biasa), `//` (pembagian dibulatkan), `%` (sisa bagi), `**` (pangkat).
- **Operator perbandingan & logika**: `==`, `!=`, `>`, `<`, `>=`, `<=`, serta `and`, `or`, `not`.
- **Input/Output**: `input()` membaca masukan pengguna sebagai `str`, dan `print()` menampilkan output ke layar.

```python
a = int(input("Masukkan angka pertama: ")) # di python, fungsi input mereturn string, jadi kita casting ke integer hasilnya (ò_óˇ)
b = int(input("Masukkan angka kedua: "))

# Operator aritmatika
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b:.2f}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")

# Operator perbandingan dan logika
print(f"Apakah {a} > {b} dan {b} > 0? {(a > b) and (b > 0)}")
```

## 3. Percabangan (`if` / `elif` / `else`)

DIbawah ini, percabangan digunakan untuk menjalankan blok kode tertentu berdasarkan nilai:
- A: 85 - 100
- B: 70 - 84
- C: 55 - 69
- D: 40 - 54
- E: < 40

```python
def konversi_nilai(nilai: float) -> str: #fungsi menerima variable nilai yang bertipe data float, lalu mereturn string
    if nilai < 0 or nilai > 100:
        return "Nilai harus 0-100"
    elif nilai >= 85:
        return "A"
    elif nilai >= 70:
        return "B"
    elif nilai >= 55:
        return "C"
    elif nilai >= 40:
        return "D"
    else:
        return "E"

daftar_nilai = [92, 78, 60, 45, 30] # ini tipe data list
for n in daftar_nilai:
    print(f"Nilai {n} -> Huruf Mutu: {konversi_nilai(n)}")
```

## 4. Perulangan (`for` & `while`): Deret Fibonacci & Bilangan Prima 1-100

- **`while` loop** cocok digunakan ketika jumlah iterasi bergantung pada kondisi tertentu (contoh: deret Fibonacci hingga batas tertentu).
- **`for` loop** cocok digunakan untuk iterasi pada rentang atau koleksi yang sudah diketahui (contoh: mengecek bilangan prima dari 1 sampai 100).

```python
# 1. Deret Fibonacci (10 suku pertama) menggunakan while loop
n = 10
a, b = 0, 1
step = 0
fib = []

while step < n:
    fib.append(a)
    a, b = b, a + b
    step += 1

print(f"Deret Fibonacci ({n} suku): {fib}")

# 2. Bilangan prima 1-100 menggunakan for loop
prime = []
for i in range(2, 101):
    is_prime = True
    for j in range(2, i):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        prime.append(i)

print(f"Bilangan prima 1-100 ({len(prime)} bilangan):\n{prime}")
```

## 5. List, Tuple, Set, Dictionary & List Comprehension

- **List** (`[]`): terurut, dapat diubah (*mutable*), boleh duplikat.
- **Tuple** (`()`): terurut, tidak dapat diubah (*immutable*), boleh duplikat.
- **Set** (`{}`): tidak terurut, elemen unik (tanpa duplikat).
- **Dictionary** (`{k: v}`): pasangan *key-value*.
- **List comprehension**: cara ringkas membuat list baru dari iterable.

```python
# List
buah = ["apel", "jeruk", "mangga", "apel"]
buah.append("pisang")
print("List buah:", buah)

# Tuple
koordinat = (-6.2088, 106.8456)
print("Tuple koordinat:", koordinat)

# Set (otomatis menghapus duplikat)
buah_unik = set(buah)
print("Set buah unik:", buah_unik)

# Dictionary
person = {
    "nama": "Raditya Alfarisi",
    "umur": 17,
}
print("Dictionary mahasiswa:", person)

# List comprehension: kuadrat bilangan genap dari 1 sampai 10
kuadrat_genap = [x ** 2 for x in range(1, 11) if x % 2 == 0]
print("List comprehension (kuadrat genap 1-10):", kuadrat_genap)
```

## 6. Function: Parameter Default, `*args` / `**kwargs`, Multiple Return, dan `lambda`

- **Parameter default**: memberi nilai bawaan jika argumen tidak diisi.
- **`*args`**: menerima jumlah argumen posisi yang bervariasi sebagai tuple.
- **`**kwargs`**: menerima jumlah argumen kata kunci yang bervariasi sebagai dictionary.
- **Multiple return**: fungsi dapat mengembalikan lebih dari satu nilai (dikemas sebagai tuple).
- **`lambda`**: fungsi anonim satu baris.

```python
# 1. Parameter default & multiple return
def hitung_persegi_panjang(panjang: float, lebar: float = 5.0):
    luas = panjang * lebar
    keliling = 2 * (panjang + lebar)
    return luas, keliling

luas, keliling = hitung_persegi_panjang(10)
print(f"Persegi panjang (10 x 5): Luas={luas}, Keliling={keliling}")

# 2. *args dan **kwargs
def ringkasan_pesanan(pelanggan: str, *item, **detail):
    print(f"Pelanggan: {pelanggan}")
    print(f"Item dipesan: {item}")
    print(f"Detail tambahan: {detail}")

ringkasan_pesanan("Raditya", "Kopi", "Roti", meja=4, catatan="Kurangi gula")

# 3. Lambda
kali_dua = lambda x: x * 2
print(f"Lambda kali_dua(7) = {kali_dua(7)}")
```

## 7. String Method: Hitung Jumlah Kata & Huruf Vokal dari Sebuah Paragraf

Menggunakan method string seperti `.lower()`, `.split()`, dan `.strip()` untuk membersihkan dan menganalisis teks.

```python
paragraf = (
    "Aku adalah dia dan dia adalah aku. "
    "Saya akan kembali ke situjuan, menjadi orang biasa."
)

# Hitung jumlah kata
kata_list = paragraf.strip().split()
jumlah_kata = len(kata_list)

# Hitung jumlah huruf vokal (a, i, u, e, o)
vokal = "aiueo"
jumlah_vokal = sum(1 for huruf in paragraf.lower() if huruf in vokal)

print(f"Paragraf:\n{paragraf}\n")
print(f"Jumlah kata  : {jumlah_kata}")
print(f"Jumlah vokal : {jumlah_vokal}")
```

## 8. Exception Handling: `try` / `except` / `finally` pada Input Angka

- **`try`**: blok kode yang berpotensi menimbulkan error.
- **`except`**: menangani error spesifik (misalnya `ValueError` jika input bukan angka).
- **`finally`**: blok yang selalu dijalankan baik terjadi error maupun tidak.

```python
def baca_angka():
    teks = input("Masukkan sebuah bilangan bulat: ")

    try:
        angka = int(teks)
        print(f"Berhasil! Angka yang Anda masukkan adalah {angka}.")
    except ValueError:
        print(f"Error: '{teks}' bukan bilangan bulat yang valid!")
    finally:
        print("Proses validasi input selesai dijalankan.")

baca_angka()
```

## 9. Modul Bawaan: `math`, `random`, dan `datetime`

- **`math`**: fungsi matematika seperti `sqrt()`, `pi`, `ceil()`, `floor()`.
- **`random`**: menghasilkan angka acak atau memilih elemen acak dari koleksi.
- **`datetime`**: bekerja dengan tanggal dan waktu saat ini maupun format tanggal.

```python
import math
import random
from datetime import datetime

# 1. math
radius = 7
luas_lingkaran = math.pi * (radius ** 2)
print(f"math: Luas lingkaran r={radius} adalah {luas_lingkaran:.2f}, akar 144 = {math.sqrt(144)}")

# 2. random
angka_acak = random.randint(1, 100)
pilihan_acak = random.choice(["Python", "JavaScript", "Go", "Rust"])
print(f"random: Angka acak (1-100) = {angka_acak}, bahasa terpilih = {pilihan_acak}")

# 3. datetime
sekarang = datetime.now()
print(f"datetime: Waktu saat ini = {sekarang.strftime('%Y-%m-%d %H:%M:%S')}")
```

## 10. Dasar OOP: `class`, `__init__`, `method`, dan Inheritance Sederhana

- **`class`**: cetak biru (*blueprint*) untuk membuat objek.
- **`__init__`**: konstruktor yang otomatis dipanggil saat objek dibuat.
- **`method`**: fungsi yang didefinisikan di dalam class.
- **Inheritance (pewarisan)**: class anak mewarisi atribut dan method dari class induk.

```python
# Class Induk
class Kendaraan:
    def __init__(self, merk: str, tahun: int):
        self.merk = merk
        self.tahun = tahun

    def info(self) -> str:
        return f"{self.merk} (Tahun {self.tahun})"

# Class Anak (Inheritance)
class Mobil(Kendaraan):
    def __init__(self, merk: str, tahun: int, jumlah_pintu: int):
        super().__init__(merk, tahun)
        self.jumlah_pintu = jumlah_pintu

    def info(self) -> str:
        return f"{super().info()} - {self.jumlah_pintu} pintu"

kendaraan1 = Kendaraan("Yamaha", 2022)
mobil1 = Mobil("Toyota Avanza", 2024, 4)

print("Kendaraan:", kendaraan1.info())
print("Mobil    :", mobil1.info())
```
